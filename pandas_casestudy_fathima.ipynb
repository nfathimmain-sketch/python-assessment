17) Create a column that categorizes movies by rating:

a) Rating < 6:poor

b) 6 ≤ Rating < 7.5: Average

c) Rating ≥ **7.5**

```python
df.loc[df['Rating'] < 6, 'Rating_Category'] = "Poor"        #a
df.loc[(df['Rating'] >= 6) & (df['Rating'] < 7.5), 'Rating_Category'] = "Average"         #b
df.loc[df['Rating'] >= 7.5, 'Rating_Category'] = "Excellent"                        #c

df[['Movie_Title', 'Rating', 'Rating_Category']].head()
```

18) Create a custom function to score movies based on:

     i) Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score
        Apply a function to dataset to assign the score

```python
# Function to calculate a movie score
def calculate_score(row):
    return row['Rating'] * 10 + row['Revenue (Millions)']

# Applying function to dataset
df['Score'] = df.apply(calculate_score, axis=1)

# Display output
df.head()
```
