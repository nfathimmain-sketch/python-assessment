Task 6

Rename columns wherever necessary so that all datasets have a common column
name for merging.

```python
excel_data.rename(columns = {'Country':'country'}, inplace=True)
xml_data.rename(columns = {'Country':'country'}, inplace=True)

print("CSV (Population):", csv_data.columns.tolist())
print("Excel (GDP):", excel_data.columns.tolist())
print("JSON (Internet Users):", json_data.columns.tolist())
print("XML (Literacy Rate):", xml_data.columns.tolist())
```

Task 7

Convert the required columns into appropriate numeric data types.

```python
csv_data['population'] = pd.to_numeric(csv_data['population'])

excel_data['GDP'] = pd.to_numeric(excel_data['GDP'])

json_data['internet_users'] = pd.to_numeric(json_data['internet_users'])

xml_data['literacy_rate'] = pd.to_numeric(xml_data['literacy_rate'])

print("CSV Data Types:\n", csv_data.dtypes, "\n")
print("Excel Data Types:\n", excel_data.dtypes, "\n")
print("JSON Data Types:\n", json_data.dtypes, "\n")
print("XML Data Types:\n", xml_data.dtypes)
```

Task 8

Merge all four datasets into one DataFrame.

Hint:

● Use the common column (country)

● Choose an appropriate join method.

```python
xml_data.rename(columns={'name': 'country'}, inplace=True)

merged_df = csv_data.merge(excel_data, on='country', how='inner').merge(json_data, on='country', how='inner').merge(xml_data, on='country', how='inner')

merged_df.head()
```

Task 9

Create the following calculated columns:

1. Internet Penetration (%)

2. GDP Per Capita

Hint:

Round the calculated values to two decimal places.

```python
# 1. Internet Penetration (%) = (internet_users / population) * 100
merged_df['internet_penetration_%'] = ((merged_df['internet_users'] / merged_df['population']) * 100).round(2)

# 2. GDP Per Capita = GDP / population
merged_df['gdp_per_capita'] = (merged_df['GDP'] / merged_df['population']).round(2)

merged_df.head()
```

Task 10

Create a SQLite database.

Store the final merged dataset into a table named
‘country_statistics’

Hint:

● Use the sqlite3 library.

● Load the DataFrame using pandas.

```python
import sqlite3

# 1. Connect to SQLite database
conn = sqlite3.connect('country_data.db')

# 2. Store the merged DataFrame into the 'country_statistics' table
merged_df.to_sql('country_statistics', conn, if_exists='replace', index=False)

# 3. Verify data by reading it back from the SQLite database
df_from_db = pd.read_sql_query("SELECT * FROM country_statistics", conn)

conn.close()
```
