Q10. Handling Missing Values
Consider the following DataFrame:

df = pd.DataFrame({
'Name': ['A', 'B', 'C', 'D'],
'Age': [21, np.nan, 23, 22],
'Score': [80, 75, np.nan, 90]
})
Write Pandas code to:
Identify the missing values.
Replace the missing Age with the mean age.
Replace the missing Score with the mean score.

```python
import pandas as pd
import numpy as np

#Create the DataFrame
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Age': [21, np.nan, 23, 22],
    'Score': [80, 75, np.nan, 90]
})

#Identify the missing values
print(df.isnull())

#Replace the missing Age with the mean age
df['Age'].fillna(df['Age'].mean(), inplace=True)

#Replace the missing Score with the mean score
df['Score'].fillna(df['Score'].mean(), inplace=True)

#Display the updated DataFrame
print(df)
```

```python
filepath = '/content/drive/MyDrive/DSA/Assessment/student_performance.csv'
students1= pd.read_csv(filepath)


print("Missing values per column:\n", students1.isnull().sum())

#Handle missing numerical values with column mean
num_cols = students1.select_dtypes(include=np.number).columns
students1[num_cols] = students1[num_cols].fillna(students1[num_cols].mean())

print("\nMissing values after cleaning:\n", students1.isnull().sum())
```

```python
students1['Average_Score'] = students1[
    ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
].mean(axis=1)

students1['Pass_Status'] = np.where(students1['Average_Score'] >= 50, 'Pass', 'Fail')

students1[['Name', 'Average_Score', 'Pass_Status']].head(10)
```

```python
#Top 5 students by Average_Score
top5 = students1.sort_values(by='Average_Score', ascending=False).head(5)
print("Top 5 students by Average_Score:")
print(top5[['Name', 'Average_Score']])
```

```python
#Department with highest average score
dept_avg = students1.groupby('Department')['Average_Score'].mean().sort_values(ascending=False)
print("\nAverage score by department:\n", dept_avg)
print("\nDepartment with highest average score:", dept_avg.idxmax())

#Average attendance by department
attendance_by_dept = students1.groupby('Department')['Attendance'].mean().round(2)
print("\nAverage attendance by department:\n", attendance_by_dept)

#ass/Fail counts
pass_fail_counts = students1['Pass_Status'].value_counts()
print("\nPass/Fail counts:\n", pass_fail_counts)
D
summary16 = students1.groupby('Department').apply(
    lambda g: pd.Series({
        'Total_Students': len(g),
        'Average_Score': round(g['Average_Score'].mean(), 2),
        'Average_Attendance': round(g['Attendance'].mean(), 2),
        'Pass_Count': (g['Pass_Status'] == 'Pass').sum(),
        'Fail_Count': (g['Pass_Status'] == 'Fail').sum()
    }),
    include_groups=False
).reset_index()
```
